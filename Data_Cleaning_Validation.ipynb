```python
missing_summary = []

for name, df in tables.items():
    for column in df.columns:
        missing_summary.append({
            'Table': name,
            'Column': column,
            'Blank Values': df[column].isnull().sum(),
            'Blank %': round(df[column].isnull().mean() * 100, 2)
        })

missing_df = pd.DataFrame(missing_summary)

display(missing_df)
```

```python
duplicate_summary = []

for name, df in tables.items():
    duplicate_summary.append({
        'Table': name,
        'Total Rows': len(df),
        'Duplicate Rows': df.duplicated().sum(),
        'Duplicate %': round(df.duplicated().mean() * 100, 2)
    })

duplicate_df = pd.DataFrame(duplicate_summary)

display(duplicate_df)
```

```python
# CUSTOMERS
customers['signup_date'] = pd.to_datetime(
    customers['signup_date'],
    errors='coerce'
)

# SESSIONS
sessions['start_time'] = pd.to_datetime(
    sessions['start_time'],
    errors='coerce'
)

# EVENTS
events['timestamp'] = pd.to_datetime(
    events['timestamp'],
    errors='coerce'
)

# Nullable integers because these columns contain NaN
events['product_id'] = events['product_id'].astype('Int64')
events['qty'] = events['qty'].astype('Int64')
events['cart_size'] = events['cart_size'].astype('Int64')

# Converting discount from object → numeric
events['discount_pct'] = pd.to_numeric(
    events['discount_pct'],
    errors='coerce'
)

# ORDERS
orders['order_time'] = pd.to_datetime(
    orders['order_time'],
    errors='coerce'
)

# REVIEWS
reviews['review_time'] = pd.to_datetime(
    reviews['review_time'],
    errors='coerce'
)
```

```python
print("Order Items:")

composite_duplicates = order_items.duplicated(
    subset=['order_id', 'product_id']
).sum()

print("Total rows:", len(order_items))
print("Duplicate order_id + product_id combinations:", composite_duplicates)
```

```python
# Foreign Key Integrity Check

print("FOREIGN KEY INTEGRITY\n")

# 1. sessions.customer_id → customers.customer_id
invalid_sessions_customer = sessions[
    ~sessions['customer_id'].isin(customers['customer_id'])
]

print("1. sessions.customer_id → customers.customer_id")
print("Invalid rows:", len(invalid_sessions_customer))


# 2. orders.customer_id → customers.customer_id
invalid_orders_customer = orders[
    ~orders['customer_id'].isin(customers['customer_id'])
]

print("\n2. orders.customer_id → customers.customer_id")
print("Invalid rows:", len(invalid_orders_customer))


# 3. events.session_id → sessions.session_id
invalid_events_session = events[
    ~events['session_id'].isin(sessions['session_id'])
]

print("\n3. events.session_id → sessions.session_id")
print("Invalid rows:", len(invalid_events_session))


# 4. events.product_id → products.product_id
invalid_events_product = events[
    events['product_id'].notna() &
    ~events['product_id'].isin(products['product_id'])
]

print("\n4. events.product_id → products.product_id")
print("Invalid rows:", len(invalid_events_product))


# 5. order_items.order_id → orders.order_id
invalid_items_order = order_items[
    ~order_items['order_id'].isin(orders['order_id'])
]

print("\n5. order_items.order_id → orders.order_id")
print("Invalid rows:", len(invalid_items_order))


# 6. order_items.product_id → products.product_id
invalid_items_product = order_items[
    ~order_items['product_id'].isin(products['product_id'])
]

print("\n6. order_items.product_id → products.product_id")
print("Invalid rows:", len(invalid_items_product))


# 7. reviews.order_id → orders.order_id
invalid_reviews_order = reviews[
    ~reviews['order_id'].isin(orders['order_id'])
]

print("\n7. reviews.order_id → orders.order_id")
print("Invalid rows:", len(invalid_reviews_order))


# 8. reviews.product_id → products.product_id
invalid_reviews_product = reviews[
    ~reviews['product_id'].isin(products['product_id'])
]

print("\n8. reviews.product_id → products.product_id")
print("Invalid rows:", len(invalid_reviews_product))
```
